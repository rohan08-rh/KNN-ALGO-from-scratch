```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
df=pd.read_csv("/content/Social_Network_Ads.csv")
df=df.iloc[:,1:]
df
```

```python
encoder=LabelEncoder()
df['Gender']=encoder.fit_transform(df['Gender'])
scaler=StandardScaler()
#df[['Age','EstimatedSalary']]=scaler.fit_transform(df[['Age','EstimatedSalary']])
#df
X=df.iloc[:,0:3].values
X=scaler.fit_transform(X)
y=df.iloc[:,-1].values
X
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
knnn=KNeighborsClassifier(n_neighbors=5)
knnn.fit(X_train,y_train)
y_pred=knnn.predict(X_test)
accuracy_score(y_test,y_pred)
```

```python
import numpy as np
from collections import Counter
class knn:
  def __init__(self,k):
    self.X_train=None
    self.y_train=None
    self.k_neigh=k
  def train(self,X,y):
    self.X_train=X
    self.y_train=y
  def predict(self,X_test):
    y_pred=[]
    for i in X_test:
      dist=[]
      for j in self.X_train:
        dist.append(self.cal_dist(i,j))
      n_neighbours=sorted(list(enumerate(dist)),key=lambda x:x[1])[0:self.k_neigh]
      #self.maj_count(n_neighbours)
      LABEL=self.maj_count(n_neighbours)
      y_pred.append(LABEL)
    return np.array(y_pred)
  def cal_dist(self,a,b):
    return np.linalg.norm(a-b)
  def maj_count(self,arr):
    votes=[]
    for i in arr:
      votes.append(self.y_train[i[0]])
    votes=Counter(votes)
    return votes.most_common(1)[0][0]
```

```python
apnaknn =knn(k=5)
apnaknn.train(X_train,y_train)
apnaknn.predict(X_test)
accuracy_score(y_test,apnaknn.predict(X_test))
```
